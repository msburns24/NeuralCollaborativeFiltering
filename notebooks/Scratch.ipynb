```python
import pandas as pd
```

```python
rating_cols = ['user', 'item', 'rating', 'ts']
base_path = 'Data/sample-ml-1m/sample-ml-1m'

train_ratings = pd.read_csv(f'{base_path}.train.rating', sep='\t', names=rating_cols)
test_ratings = pd.read_csv(f'{base_path}.test.rating', sep='\t', names=rating_cols)
test_negatives = pd.read_csv(f'{base_path}.test.negative', sep='\t', header=None)
test_negatives.head()
```

```python
print('Highest item ID in train ratings:', train_ratings['item'].max())
print('Highest item ID in test ratings:', test_ratings['item'].max())
print('Highest item ID in test negatives:', test_negatives.drop(columns=0).max(axis=None))
```
